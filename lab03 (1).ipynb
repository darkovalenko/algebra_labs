### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
from collections import Counter

for n in [100, 173, 530, 746, 911, 1024]:
    print("=" * 70)
    print(f"n = {n},  розклад: {factor(n)}")
    G = Zmod(n)                      
    divs = divisors(n)               
    print("Дільники:", divs)

    cnt = Counter(G(a).additive_order() for a in range(n))

    print(f"{'d':>6} {'кількість':>10} {'phi(d)':>8}  збігається?")
    ok_all = True
    for d in divs:
        c, ph = cnt[d], euler_phi(d)
        ok_all &= (c == ph)
        print(f"{d:>6} {c:>10} {ph:>8}  {c == ph}")
    print("Усі кількості збігаються з phi(d):", ok_all)

    subgroups = {d: sorted(set((n // d) * k % n for k in range(d))) for d in divs}
    for d in divs:
        H = subgroups[d]
        shown = H if len(H) <= 12 else H[:6] + ['...'] + H[-3:]
        print(f"  підгрупа порядку {d:>4}: <{n // d}> = {shown}")

    C = CyclicPermutationGroup(n)
    sage_count = len(C.subgroups())
    print("Кількість підгруп (явна побудова):", len(subgroups))
    print("Кількість підгруп (Sage, subgroups()):", sage_count)
    print("tau(n) =", number_of_divisors(n),
          "| збігається:", len(subgroups) == sage_count == number_of_divisors(n))
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
from collections import Counter

results = {}
for n in [5, 6, 10, 12, 16, 17, 20]:
    D = DihedralGroup(n)
    assert D.order() == 2 * n
    cnt = Counter(g.order() for g in D)
    results[n] = cnt

    print("=" * 70)
    print(f"D_{n}, |D_n| = {2*n}, дільники порядку: {divisors(2*n)}")
    print("Кількість елементів порядку 2:", cnt[2],
          "| теорія:", n + 1 if n % 2 == 0 else n)
    for d in divisors(2 * n):
        if d > 2:
            print(f"  порядок d = {d:>3}: {cnt[d]:>3} елемент(ів)",
                  "" if cnt[d] > 0 else "  <-- немає елементів такого порядку")
    missing = [d for d in divisors(2 * n) if d < 2 * n and cnt[d] == 0]
    print("Дільники d < 2n, для яких НЕМАЄ елемента порядку d:", missing)

print("\nПеревірка формули: #(порядок d>2) = phi(d), якщо d | n, інакше 0")
for n, cnt in results.items():
    ok = all(cnt[d] == (euler_phi(d) if n % d == 0 else 0)
             for d in divisors(2 * n) if d > 2)
    print(f"  n = {n:>2}: {ok}")
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
n = 1102158078129785185997827239806

F = factor(n)
print("n =", n)
print("Факторизація:", F)

G = Zmod(n)
phi_n = euler_phi(n)
print("Група циклічна (Sage):", G.multiplicative_group_is_cyclic())
def is_cyclic_unit_group(n):
    if n in (1, 2, 4):
        return True
    f = factor(n)
    if len(f) == 1:
        return f[0][0] != 2          # p^k, p непарне
    if len(f) == 2 and f[0][0] == 2 and f[0][1] == 1:
        return True                  # 2 * p^k
    return False
print("Група циклічна (теорія):", is_cyclic_unit_group(n))
print("Порядок групи phi(n) =", phi_n)
print("Факторизація phi(n):", factor(phi_n))

prime_divs = [q for q, _ in factor(phi_n)]
def is_generator(a):
    if gcd(a, n) != 1:
        return False
    x = G(a)
    return all(x ** (phi_n // q) != 1 for q in prime_divs)

a = next(a for a in range(2, 10**6) if is_generator(a))
print("Найменший твірний елемент a =", a)
print("Перевірка Sage primitive_root(n):", primitive_root(n))

print("ord(a) =", G(a).multiplicative_order())
print("ord(a) == phi(n):", G(a).multiplicative_order() == phi_n)

set_random_seed(2024)
gs = []
while len(gs) < 3:
    g = G.random_element()
    if gcd(g.lift(), n) == 1:
        gs.append(g)
print("Випадкові елементи:", gs)
base = G(a)
for g in gs:
    k = discrete_log(g, base, ord=phi_n)  
    print(f"g = {g}\n  k = {k}\n  перевірка a^k == g: {base ** k == g}")
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:
1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?

```sage
A = matrix(ZZ, [[1, 0, 1], [0, 1, 0], [0, 0, 1]])
B = matrix(ZZ, [[1, 0, 0], [0, 1, 1], [0, 0, 1]])
print("det A =", A.det(), ", det B =", B.det())

print("AB == BA:", A * B == B * A)

print("A^k для k = 1..5:")
for k in range(1, 6):
    print(f"  k={k}:", (A ** k).rows())
print("A^k = I лише при k = 0 => A має нескінченний порядок => G нескінченна")

try:
    Gm = MatrixGroup([A, B])
    print("G.is_finite() (Sage/GAP):", Gm.is_finite())
except Exception as e:
    print("MatrixGroup над ZZ: операція недоступна ->", type(e).__name__)

var('a_, b_')
M = matrix(SR, [[1, 0, a_], [0, 1, b_], [0, 0, 1]])
ok = all(A**a * B**b == matrix(ZZ, [[1, 0, a], [0, 1, b], [0, 0, 1]])
         for a in range(-6, 7) for b in range(-6, 7))
print("A^a B^b = [[1,0,a],[0,1,b],[0,0,1]] для |a|,|b| <= 6:", ok)

I3 = identity_matrix(ZZ, 3)
orders = {}
for a in range(-5, 6):
    for b in range(-5, 6):
        g = A**a * B**b
        if g == I3:
            o = 1
        else:
            o = Infinity
            assert all(g**k != I3 for k in range(1, 51))
        orders[(a, b)] = o
from collections import Counter
print("Порядки елементів (вибірка |a|,|b| <= 5):", Counter(orders.values()))

print("\nG абелева, нескінченна, всі нетривіальні елементи мають нескінченний порядок.")
print("Рангом 2 => G НЕ циклічна (Z^2 не породжується одним елементом).")

phi_map = lambda a, b: A**a * B**b
hom = all(phi_map(a1, b1) * phi_map(a2, b2) == phi_map(a1 + a2, b1 + b2)
          for a1 in range(-3, 4) for b1 in range(-3, 4)
          for a2 in range(-3, 4) for b2 in range(-3, 4))
print("Гомоморфізм (a,b) -> A^a B^b:", hom)
print("Отже, G ≅ Z × Z = Z^2 (вільна абелева група рангу 2).")
```
