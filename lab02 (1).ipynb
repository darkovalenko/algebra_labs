### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
print("=== 2. Група Z10 ===")
G1 = CyclicPermutationGroup(10)
elems = G1.list()
a, b = elems[1], elems[2]
print("1. Побудовано групу Z10")
print("2. Вибрані елементи a =", a, ", b =", b)
print("   Добуток a * b =", a * b)
print("3. Обернені елементи a^-1 =", a^-1, ", b^-1 =", b^-1)
print("   Перевірка:", a * a^-1 == G1.one() and b * b^-1 == G1.one())
H1 = G1.subgroup([a, b])
print("4. Порядок підгрупи <a, b>:", H1.order())
print("5. Елементи, де a^-1 = a:", [x for x in G1 if x^-1 == x])
if not G1.is_abelian():
    pair = next((x, y) for x in G1 for y in G1 if x * y != y * x)
    print("6. Елементи, що не комутують:", pair[0], "*", pair[1], "!=", pair[1], "*", pair[0])
else:
    print("6. Група є абелевою (усі елементи комутують)")
```

```sage
G = Zmod(10).unit_group()
a, b = G.random_element(), G.random_element()
print(f"--- Група Z_10^* ---\nЕлементи: a = {a}, b = {b}")
print(f"Добуток a*b = {a*b}")
print(f"Обернені: a^-1 = {a^-1}, b^-1 = {b^-1}")
print(f"Перевірка (a * a^-1 == e): {a*a^-1 == G.identity()}")
H = G.subgroup([a, b])
print(f"Порядок підгрупи <a, b>: {H.order()}")
inv_self = [g for g in G if g == g^-1]
print(f"Елементи, що обернені самі до себе: {inv_self}")
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
print("=== 3. Підгрупи групи Z24 ===")
G1 = CyclicPermutationGroup(24)
subgroups1 = G1.subgroups()
print("1. Побудовано групу Z24")
print("2. Список усіх підгруп:", subgroups1)
iso1 = []
for sg in subgroups1:
    if not any(sg.is_isomorphic(c[0]) for c in iso1):
        iso1.append([sg])
print("3. Кількість підгруп з точністю до ізоморфізму:", len(iso1))
orders1 = [sg.order() for sg in subgroups1]
print("4. Порядки всіх підгруп:", orders1)
print("5. Перевірка (порядок підгрупи ділить порядок групи):", all(G1.order() % o == 0 for o in orders1))
```

```sage
G = QuaternionGroup()
subs = G.subgroups()
iso_subs = G.conjugacy_classes_subgroups()
print("--- Група Q_8 ---")
print("Кількість підгруп:", len(subs))
print("Кількість підгруп (з точністю до ізоморфізму):", len(iso_subs))
orders = [H.order() for H in subs]
print("Порядки підгруп:", set(orders))
print("Чи ділить порядок кожної підгрупи порядок групи?", all(G.order() % o == 0 for o in orders))
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
G = DihedralGroup(4)
C = G.cayley_graph()
print("D_4: Степені вершин:", set(C.degree()))
print("Вершинно-транзитивний?", C.is_vertex_transitive())
print("Чи містить гамільтонів цикл?", C.is_hamiltonian())
C.plot().show()
```

```sage
G = QuaternionGroup()
C = G.cayley_graph()
print("Q_8: Степені вершин:", set(C.degree()))
print("Вершинно-транзитивний?", C.is_vertex_transitive())
print("Чи містить гамільтонів цикл?", C.is_hamiltonian())
C.plot().show()
```
