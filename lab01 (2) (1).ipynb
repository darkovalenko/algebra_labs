### 1. Арифметичні операції

а) додайте 25 і 17;<br>
б) помножте 7 на 9;<br>
в) поділіть 144 на 12;<br>
г) знайдіть залишок від ділення 156 на 7;<br>
д) поділіть 156 націло на 7;<br>
е) розкладіть число 2024 на прості множники.

```sage
# Ваш код тут:
print("а) 25 + 17 =", 25 + 17)
print("б) 7 * 9 =", 7 * 9)
print("в) 144 / 12 =", 144 / 12)
print("г) Залишок від ділення 156 на 7 =", 156 % 7)
print("д) Ділення 156 націло на 7 =", 156 // 7)
print("е) Розклад числа 2024 на прості множники =", factor(2024))
```


### 2. Степені, корені та стандартні функції

###  3. Символічні обчислення та алгебра виразів
а) оголосіть змінні $x=10$ і $y=15$ та обчисліть їхню суму і добуток;<br>
б) оголосіть символічну змінну $z$ і обчисліть вираз $3z^2 + 2z - 5$;<br>
в) спростіть вираз $2x^2 + 4x - 2x^2 + x$;<br>
г) розкрийте дужки у виразі $(x + 2)(x - 3)$;<br>
д) розкладіть на множники тричлен $x^2 - 9x + 20$;<br>
е) розв'яжіть рівняння $x^2 - 4x + 3 = 0$.<br>

```sage
# Ваш код тут:
x_val = 10
y_val = 15
print("а) Сума =", x_val + y_val, "| Добуток =", x_val * y_val)

var('z')
expr_z = 3*z^2 + 2*z - 5
print("б) Вираз із z =", expr_z)

var('x') 
expr_x = 2*x^2 + 4*x - 2*x^2 + x
print("в) Спрощений вираз =", expr_x.simplify_full())

print("г) Розкриті дужки =", expand((x + 2)*(x - 3)))

print("д) Розклад на множники =", factor(x^2 - 9*x + 20))

print("е) Корені рівняння =", solve(x^2 - 4*x + 3 == 0, x))
```

### 4. Похідні, інтеграли та графіки функцій

а) побудуйте графік функції $f(x) = x^2 - 4x + 3$ на відрізку $[-5, 5]$; <br>
б) знайдіть похідну функції $f(x) = x^3 + 5x^2 - 2x + 7$; <br>
в) знайдіть невизначений інтеграл функції $2x^3 - x^2 + 5$; <br>
г) обчисліть визначений інтеграл $\int_{1}^{4} x^2\, dx$; <br>
д) оголосіть функцію $f(x) = 3x^2 - 2x + 1$ та обчисліть $f(5)$; <br>
е) побудуйте на одній площині графіки $f(x)=x^2$ і $g(x)=-x^2+5$ на $[-3,3]$.


```sage
# Ваш код тут:
var('x') 
f1 = x^2 - 4*x + 3
plot1 = plot(f1, (x, -5, 5), title="Графік f(x) = x^2 - 4x + 3")
plot1.show()

f2 = x^3 + 5*x^2 - 2*x + 7
print("б) Похідна =", diff(f2, x))
f3 = 2*x^3 - x^2 + 5
print("в) Невизначений інтеграл =", integral(f3, x))

print("г) Визначений інтеграл =", integral(x^2, x, 1, 4))

f4(x) = 3*x^2 - 2*x + 1
print("д) Значення функції f(5) =", f4(5))

p_f = plot(x^2, (x, -3, 3), color='blue', legend_label='f(x) = x^2')
p_g = plot(-x^2 + 5, (x, -3, 3), color='red', legend_label='g(x) = -x^2 + 5')
(p_f + p_g).show(title="Два графіки на одній площині")
```

### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
# Ваш код тут:
n = 7  
print("а) Перевірка числа", n, ":")
if n > 0:
    print("Число додатне")
elif n < 0:
    print("Число від'ємне")
else:
    print("Число дорівнює нулю")
    
print("-" * 20)


print("б) Цикл від 1 до 10:")
for i in range(1, 11):
    print(i, end=" ") 
print("\n" + "-" * 20) 

def check_prime(number):
    if is_prime(number):
        return f"Число {number} є простим."
    else:
        return f"Число {number} не є простим."

print("в) Робота функції:")
print(check_prime(17))
print(check_prime(20))
```
